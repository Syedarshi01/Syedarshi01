1. Creating a database in memory:

  conn.cursor() is not a mandatory function to use in sql in general but in python, conn.cursor() creates a cursor object through which we execute sql commands.

  Connecting to a database.
  

```python
import sqlite3

#create a memory for the database
conn = sqlite3.connect(":memory:")

#create a cursor to execute sql commands
cursor = conn.cursor()
print("database created successfully")
```

2. Creating my first table:

```python
cursor.execute("""
CREATE TABLE One(
    id INTEGER,
    name TEXT,
    age INTEGER,
    marks REAL
)
""")

print("table created!")
```

3. Inserting data into the table:

```python
cursor.execute("""

INSERT INTO One(id, name, age, marks)
VALUES
  (1, 'Arshiya', 21, 85),
  (2, 'Ayesha', 23, 95),
  (3, 'Shraya', 22, 94),
  (4, 'Ammulu', 24, 83)
""")

conn.commit()

print("Students record created successfully!")
```

4. Retrieving the records:

```python
cursor.execute("SELECT * FROM One")

rows = cursor.fetchall()

for row in rows:
  print(row)
```

5. Displaying one category by retrieving it separately:

```python
cursor.execute("SELECT name FROM One")

rows = cursor.fetchall()

for row in rows:
  print(row)

cursor.execute("SELECT age FROM One")

rows = cursor.fetchall()

for row in rows:
  print(row)
```
